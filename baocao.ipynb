### 1. Thông tin nguồn dữ liệu
**Website:** https://books.toscrape.com/ (Trang web cho phép luyện tập cào dữ liệu).  
**Số lượng dữ liệu:** Lấy 10 trang, mỗi trang 20 cuốn -> Tổng cộng 200 cuốn sách.  
**Công cụ sử dụng:** Thư viện requests để lấy HTML và BeautifulSoup để bóc tách dữ liệu.  
**Các cột dữ liệu lấy được:** Tên sách, giá, đánh giá, tình trạng, thể loại, link.  


```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('data_sach.csv')
print("Số lượng dòng và cột:", df.shape)
df.head()
```

### 2. Thống kê cơ bản
Kiểm tra xem dữ liệu có bị thiếu (null) không và xem thống kê giá.

```python
print("Số giá trị rỗng trong mỗi cột:")
print(df.isnull().sum())
```

```python
df[['gia', 'danhgia']].describe()
```

Nhận xét: Giá sách trung bình tầm 34 - 35 bảng, cao nhất là khoảng gần 60 bảng và thấp nhất khoảng 10 bảng. Dữ liệu thu thập đủ 200 cuốn và không bị rỗng ở cột nào.

### 3. Trực quan hóa dữ liệu
Vẽ vài biểu đồ cơ bản để xem dữ liệu phân bố như nào.

```python
plt.figure(figsize=(6, 4))
df['danhgia'].value_counts().sort_index().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Số lượng sách theo mức đánh giá')
plt.xlabel('Số sao')
plt.ylabel('Số lượng')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.hist(df['gia'], bins=15, color='lightgreen', edgecolor='black')
plt.title('Phân bố giá tiền của các cuốn sách')
plt.xlabel('Giá')
plt.ylabel('Số lượng')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
toptheloai = df['theloai'].value_counts().head(10)
toptheloai.sort_values().plot(kind='barh', color='salmon', edgecolor='black')
plt.title('Top 10 thể loại xuất hiện nhiều nhất')
plt.xlabel('Số lượng sách')
plt.show()
```
